# 1-3-7 Pandas 樞紐分析表 (pivot_table)

樞紐分析表可將資料依「列」與「欄」分組，再用聚合函數（sum、count、mean…）彙整數值，類似 Excel 的樞紐分析表。

**重要參數**
- `index`：列索引（分組依據）
- `columns`：欄索引（分類依據）
- `values`：要統計的欄位
- `aggfunc`：聚合函數

## 1. 匯入套件並讀取資料

```python
import pandas as pd

df = pd.read_csv('mock_orders.csv')
df.head()
```

```python
# 查看欄位名稱
df.columns
```

## 2. 樞紐分析表：各年度、各類別的銷售總和

列：年度 (`Year`)，欄：商品類別 (`Category`)，值：`Sales` 的總和。

```python
sum_table = pd.pivot_table(
    df,                  # 原始資料（DataFrame）
    index="Year",        # 列索引：依「年度」分組
    columns="Category",  # 欄索引：依「類別」分類
    values="Sales",      # 用來統計的欄位
    aggfunc="sum"        # 聚合函數：每個 年度×類別 的總和
)

sum_table
```

## 3. 樞紐分析表：各年度、各類別的訂單數量

把 `aggfunc` 換成 `count`，即可計算筆數。

```python
pivot_count = pd.pivot_table(
    df,
    index="Year",
    columns="Category",
    values="Sales",
    aggfunc="count"
)

pivot_count
```

## 4. 延伸練習

試試其他參數：`aggfunc="mean"`、`fill_value=0`（缺值補 0）、`margins=True`（加總計）。

```python
pd.pivot_table(
    df,
    index="Year",
    columns="Category",
    values="Sales",
    aggfunc="mean",
    fill_value=0,
    margins=True,
    margins_name="總計"
)
```
