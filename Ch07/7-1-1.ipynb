# 7-1-1維度與指標

> 請以此檔所在章節資料夾作為工作目錄執行，才能讀到相對路徑的資料檔。

```python
import pandas as pd
import numpy as np
```

```python
# 1. 載入資料集
df = pd.read_csv('business_dataset.csv')
df.columns
```

```python
# 維度
dimensions = [
    'region',
    'store',
    'category',
    'channel',
    'customer_type'
]
```

```python
# 指標
metrics = [
    'revenue',
    'profit',
    'quantity'
]
```

```python
# groupby 本質上就是把維度當成觀察角度，對指標進行聚合。
# 觀察不同區域的營收
olap = df.groupby('region')['revenue'].sum()
print(olap)
# 觀察不同渠道利潤
olap = df.groupby('channel')['profit'].sum()
print(olap)
# 觀察不同站面的平均利潤
olap = df.groupby('store')['profit'].mean()
print(olap)
```
